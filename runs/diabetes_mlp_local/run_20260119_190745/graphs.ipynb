```python
import json
import pandas as pd
import matplotlib.pyplot as plt

# Path to your history.json
history_path = "history.json"

with open(history_path, "r") as f:
    history = json.load(f)

df = pd.DataFrame(history)

fig, ax1 = plt.subplots(figsize=(8, 5))

# Train / validation loss
ax1.plot(df["epoch"], df["train_loss"], label="Train Loss", linewidth=2)
ax1.plot(df["epoch"], df["val_loss"], label="Val Loss", linewidth=2)
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Loss")
ax1.grid(True, alpha=0.3)
ax1.legend(loc="upper right")

# Learning rate on secondary axis (log scale)
ax2 = ax1.twinx()
ax2.plot(df["epoch"], df["lr"], linestyle="--", linewidth=2)
ax2.set_ylabel("Learning Rate")
ax2.set_yscale("log")

plt.title("Training History")
plt.show()
```
